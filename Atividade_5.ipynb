1. Sistemas lineares na forma matricial

a) $$\begin{cases}2x + y = 7\\x - 3y = -5\end{cases}$$

Matriz dos coeficientes:
$$A = \begin{bmatrix}2 & 1 \\ 1 & -3\end{bmatrix}$$

Vetor de incógnitas:
$$x = \begin{bmatrix}x \\ y\end{bmatrix}$$

Vetor independente:
$$b = \begin{bmatrix}7 \\ -5\end{bmatrix}$$

Portanto, Ax = b vai ser:
$$\begin{bmatrix}2 & 1 \\ 1 & -3\end{bmatrix}\begin{bmatrix}x \\ y\end{bmatrix} = \begin{bmatrix}7 \\ -5\end{bmatrix}$$

b) $$\begin{cases}x + 2y - z = 2\\3x - y + 2z = -5\\2x + y - z = 7\end{cases}$$

Matriz dos coeficientes:
$$A = \begin{bmatrix}1 & 2 & -1 \\ 3 & -1 & 2 \\ 2 & 1 & 1\end{bmatrix}$$

Vetor de incógnitas:
$$b = \begin{bmatrix}x \\ y \\ z\end{bmatrix}$$

Vetor independente:
$$b = \begin{bmatrix}2 \\ 9 \\ 7\end{bmatrix}$$

Portanto, Ax = b vai ser:
$$\begin{bmatrix}1 & 2 & -1 \\ 3 & -1 & 2 \\ 2 & 1 & 1\end{bmatrix}\begin{bmatrix}x \\ y \\ z\end{bmatrix} = \begin{bmatrix}2 \\ 9 \\ 7\end{bmatrix}$$

2. Método de eliminação de Gauss

$$\begin{cases}2x + y - z = 8\\-3x - y + 2z = -11\\-2x + y + 2z = -3\end{cases}$$

Matriz aumentada:
$$\begin{bmatrix}2 & 1 & -1 & | & 8\\ -3 & -1 & 2 & | & -11 \\ -2 & 1 & 2 & | & -3\end{bmatrix}$$

Passo 1 - zerar os elementos abaixo do primeiro pivô

O primeiro pivô é 2

Para a segunda linha: $L_{2} ← 2L_{2} + 3L_{1}$

Para a terceira linha: $L_{3} ← L_{3} + L_{1}$

Assim:

$$\begin{bmatrix}2 & 1 & -1 & | & 8\\ 0 & 1 & 1 & | & -10 \\ 0 & 2 & 1 & | & 5\end{bmatrix}$$

Passo 2 - zerar o elemento abaixo do segundo pivô

O segundo pivô é 1

Fazemos: $L_{3} ← L_{3} - 2L_{2}$

Obtemos:

$$\begin{bmatrix}2 & 1 & -1 & | & 8\\ 0 & 1 & 1 & | & -10 \\ 0 & 0 & -1 & | & 25\end{bmatrix}$$

Substituição regressiva

Da terceira linha: $$-z = 25$$ Portanto: $$z = -25$$

Substituindo na segunda equação: $$y + z = -10$$ $$y - 25 = -10$$ $$y = 15$$

Substituindo na primeira:
$$2x + y - z = 8$$
$$2x + 15 - (-25) = 8$$
$$2x + 40 = 8$$
$$2x = -32$$
$$x = -16$$

Portanto:
$$x = -16, y = 15, z = -25$$

## Parte 2 - Propriedades e interpretação

3. Números de soluções

a) $$\begin{cases}x + y = 3\\2x + 2y = 6\end{cases}$$
Matriz aumentada:
$$\begin{bmatrix}1 & 1 & | & 3\\ 2 & 2 & | & 6\end{bmatrix}$$
Fazendo:
$$L_{2} ← L_{2} - 2L_{1}$$
Temos:
$$A = \begin{bmatrix}1 & 1 & | & 3\\ 0 & 0 & | & 0\end{bmatrix}$$
Logo, temos que:
$$x + y = 3$$
Portanto existem infinitas soluções, podendo escrever como: $x = 3 - y$

b) $$\begin{cases}x + y = 3\\2x + 2y = 7\end{cases}$$
Matriz aumentada:
$$\begin{bmatrix}1 & 1 & | & 3\\ 2 & 2 & | & 7\end{bmatrix}$$
Fazendo:
$$L_{2} ← L_{2} - 2L_{1}$$
Temos:
$$A = \begin{bmatrix}1 & 1 & | & 3\\ 0 & 0 & | & 1\end{bmatrix}$$
Logo, temos que:
$$0 = 1$$
Portanto não existem soluções

c) $$\begin{cases}x + y = 3\\2x - y = 0\end{cases}$$
Matriz aumentada:
$$\begin{bmatrix}1 & 1 & | & 3\\ 2 & -1 & | & 0\end{bmatrix}$$
Fazendo:
$$L_{2} ← L_{2} - 2L_{1}$$
Temos:
$$A = \begin{bmatrix}1 & 1 & | & 3\\ 0 & -3 & | & -6\end{bmatrix}$$
Logo, temos que:
$$-3y = -6$$
$$y = 2$$
$$x + y = 3$$
$$x + 2 = 3$$
$$x = 1$$
Portanto existe uma única solução, $(x,y) = (1,2)$

## Parte 3 - Aplicação

4. Eliminação de Gauss

```python
import numpy as np

def eliminacao_gauss(A, b):
    # Faz cópias para não alterar os dados originais
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)

    n = len(b)

    # Monta a matriz aumentada [A | b]
    matriz = np.column_stack((A, b))

    print("Matriz aumentada inicial:")
    print(matriz)

    # Eliminação de Gauss
    for k in range(n - 1):

        # Verifica se o pivô é zero
        if matriz[k, k] == 0:
            raise ValueError("Pivô zero. O método sem pivoteamento não pode continuar.")

        # Elimina os elementos abaixo do pivô
        for i in range(k + 1, n):

            fator = matriz[i, k] / matriz[k, k]

            matriz[i, k:] = matriz[i, k:] - fator * matriz[k, k:]

        print(f"\nApós a eliminação da coluna {k + 1}:")
        print(matriz)

    # Substituição regressiva
    x = np.zeros(n)

    for i in range(n - 1, -1, -1):

        soma = 0

        for j in range(i + 1, n):
            soma += matriz[i, j] * x[j]

        x[i] = (matriz[i, n] - soma) / matriz[i, i]

    return x
```

```python
A = [
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
]

b = [8, -11, -3]

solucao_gauss = eliminacao_gauss(A, b)

print("\nSolução pelo método de Gauss:")
print(solucao_gauss)
```

```python
solucao_numpy = np.linalg.solve(A, b)

print("Solução usando numpy.linalg.solve:")
print(solucao_numpy)

print("\nDiferença entre os métodos:")
print(solucao_gauss - solucao_numpy)
```

O método implementado manualmente utilizando eliminação de Gauss encontrou a mesma solução que numpy.linalg.solve. Portanto, para esse sistema, os dois métodos apresentam resultados equivalentes, com diferença numérica praticamente nula.

Modelo de balanço térmico

### a) Resolução manual

$$\begin{bmatrix}4 & -1 & 0 \\ -1 & 4 & -1 \\ 0 & -1 & 3\end{bmatrix}\begin{bmatrix}T_{1} \\ T_{2} \\ T_{3}\end{bmatrix} = \begin{bmatrix}15 \\ 10 \\ 10\end{bmatrix}$$
Matriz aumentada:
$$\begin{bmatrix}4 & -1 & 0 & | & 15\\ -1 & 4 & -1 & | & 10 \\ 0 & -1 & 3 & | & 10\end{bmatrix}$$
Eliminando o elemento abaixo do primeiro pivô:
$$L_{2} ← 4L_{2} + L_{1}$$
Então:
$$\begin{bmatrix}4 & -1 & 0 & | & 15\\ 0 & 15 & -4 & | & 55 \\ 0 & -1 & 3 & | & 10\end{bmatrix}$$
Agora eliminando o -1:
$$L_{3} ← 15L_{3} + L_{2}$$
Portato:
$$\begin{bmatrix}4 & -1 & 0 & | & 15\\ 0 & 15 & -4 & | & 55 \\ 0 & 0 & 41 & | & 205\end{bmatrix}$$
Fazendo a substituição regressiva:
$$41T_{3} = 205$$
Assim: $$T_{3} = 5$$
Da segunda linha:
$$15T_{2} - 4T_{3} = 55$$
$$15T_{2} - 20 = 55$$
$$15T_{2} = 75$$
$$T_{2} = 5$$
Da primeira:
$$4T_{1} - T_{2} = 15$$
$$4T_{1} - 5 = 15$$
$$4T_{1} = 20$$
$$T_{1} = 5$$
Portanto: $$T_{1} = T_{2} = T_{3} = 5$$

### b) Implementação em Python

```python
A = [
    [4, -1, 0],
    [-1, 4, -1],
    [0, -1, 3]
]

b = [15, 10, 10]

solucao = eliminacao_gauss(A, b)

print("\nTemperaturas calculadas:")
print(f"T1 = {solucao[0]:.2f}")
print(f"T2 = {solucao[1]:.2f}")
print(f"T3 = {solucao[2]:.2f}")
```

```python
solucao_numpy = np.linalg.solve(A, b)

print("Solução usando NumPy:")
print(solucao_numpy)
```
